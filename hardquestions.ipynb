```python
duplicates = df[df.duplicated(
    subset=['question', 'role'],
    keep=False
)]

print("Duplicate rows:", len(duplicates))
display(duplicates)
```

```python
dff = df[df['answer'].duplicated(keep=False)]

print("Total duplicate answer rows:", len(dff))
display(dff)
```

```python
df['role'].nunique()
```

```python
selected_roles = [
    "Application Security (AppSec) Engineer",
    "Data Analyst",
    "Embedded Systems Developer",
    "Cloud Infrastructure Engineer",
    "Test Automation Engineer",
    "Scrum Master / Agile Coach",
    "Database Administrator (DBA)",
    "Sales / Solutions Engineer",
    "Data Scientist",
    "Network Engineer"
]
```

```python
df = df[df['role'].isin(selected_roles)].reset_index(drop=True)
```
